```python
import numpy as np

# ============================================================
# CUSTOMER VELOCITY — FINAL CORRECT VERSION
# ============================================================

customer_velocity_base = features_df[
    ["TRANSACTION_ID", "CUSTOMER_ID", "TX_DATETIME"]
].copy()

customer_velocity_base = customer_velocity_base.sort_values(
    ["CUSTOMER_ID", "TX_DATETIME", "TRANSACTION_ID"]
)

def calculate_velocity(group):
    group = group.sort_values(
        ["TX_DATETIME", "TRANSACTION_ID"]
    ).copy()

    timestamps = group["TX_DATETIME"].values
    n = len(timestamps)

    # Convert timestamps to nanoseconds for searchsorted
    timestamps_ns = timestamps.astype("datetime64[ns]").astype(np.int64)

    current_indices = np.arange(n)

    for window, column in [
        (5 * 60, "CUSTOMER_TX_COUNT_5M"),
        (60 * 60, "CUSTOMER_TX_COUNT_1H"),
        (24 * 60 * 60, "CUSTOMER_TX_COUNT_24H")
    ]:

        window_ns = window * 1_000_000_000

        # Left boundary: current time - window
        left_boundary = timestamps_ns - window_ns

        # First transaction >= left boundary
        left_indices = np.searchsorted(
            timestamps_ns,
            left_boundary,
            side="left"
        )

        # First transaction >= current timestamp
        # This excludes transactions at the exact current timestamp.
        right_indices = np.searchsorted(
            timestamps_ns,
            timestamps_ns,
            side="left"
        )

        group[column] = (
            right_indices - left_indices
        ).astype(int)

    return group


customer_velocity_features = (
    customer_velocity_base
    .groupby("CUSTOMER_ID", group_keys=False)
    .apply(calculate_velocity)
    .reset_index(drop=True)
)

customer_velocity_features = customer_velocity_features[
    [
        "TRANSACTION_ID",
        "CUSTOMER_TX_COUNT_5M",
        "CUSTOMER_TX_COUNT_1H",
        "CUSTOMER_TX_COUNT_24H"
    ]
]

print("CUSTOMER VELOCITY — FINAL CHECK")
print("=" * 50)

print("Rows:",
      len(customer_velocity_features))

print("Unique transaction IDs:",
      customer_velocity_features["TRANSACTION_ID"].nunique())

print("\nNaN counts:")
print(customer_velocity_features.isna().sum())

print("\nMaximum values:")
print(
    customer_velocity_features[
        [
            "CUSTOMER_TX_COUNT_5M",
            "CUSTOMER_TX_COUNT_1H",
            "CUSTOMER_TX_COUNT_24H"
        ]
    ].max()
)

print("\nNon-zero counts:")
print(
    (customer_velocity_features[
        [
            "CUSTOMER_TX_COUNT_5M",
            "CUSTOMER_TX_COUNT_1H",
            "CUSTOMER_TX_COUNT_24H"
        ]
    ] > 0).sum()
)
```

```python
# ============================================================
# TERMINAL VELOCITY FEATURES
# ============================================================

terminal_velocity_base = features_df[
    ["TRANSACTION_ID", "TERMINAL_ID", "TX_DATETIME"]
].copy()

terminal_velocity_base = terminal_velocity_base.sort_values(
    ["TERMINAL_ID", "TX_DATETIME", "TRANSACTION_ID"]
)

def calculate_terminal_velocity(group):
    group = group.sort_values(
        ["TX_DATETIME", "TRANSACTION_ID"]
    ).copy()

    timestamps_ns = (
        group["TX_DATETIME"]
        .values
        .astype("datetime64[ns]")
        .astype(np.int64)
    )

    for window, column in [
        (5 * 60, "TERMINAL_TX_COUNT_5M"),
        (60 * 60, "TERMINAL_TX_COUNT_1H"),
        (24 * 60 * 60, "TERMINAL_TX_COUNT_24H")
    ]:

        window_ns = window * 1_000_000_000

        left_boundary = timestamps_ns - window_ns

        left_indices = np.searchsorted(
            timestamps_ns,
            left_boundary,
            side="left"
        )

        right_indices = np.searchsorted(
            timestamps_ns,
            timestamps_ns,
            side="left"
        )

        group[column] = (
            right_indices - left_indices
        ).astype(int)

    return group


terminal_velocity_features = (
    terminal_velocity_base
    .groupby("TERMINAL_ID", group_keys=False)
    .apply(calculate_terminal_velocity)
    .reset_index(drop=True)
)

terminal_velocity_features = terminal_velocity_features[
    [
        "TRANSACTION_ID",
        "TERMINAL_TX_COUNT_5M",
        "TERMINAL_TX_COUNT_1H",
        "TERMINAL_TX_COUNT_24H"
    ]
]

print("TERMINAL VELOCITY CHECK")
print("=" * 50)

print("Rows:",
      len(terminal_velocity_features))

print("Unique transaction IDs:",
      terminal_velocity_features["TRANSACTION_ID"].nunique())

print("\nNaN counts:")
print(terminal_velocity_features.isna().sum())

print("\nMaximum values:")
print(
    terminal_velocity_features[
        [
            "TERMINAL_TX_COUNT_5M",
            "TERMINAL_TX_COUNT_1H",
            "TERMINAL_TX_COUNT_24H"
        ]
    ].max()
)

print("\nNon-zero counts:")
print(
    (terminal_velocity_features[
        [
            "TERMINAL_TX_COUNT_5M",
            "TERMINAL_TX_COUNT_1H",
            "TERMINAL_TX_COUNT_24H"
        ]
    ] > 0).sum()
)
```

```python
features_df = features_df.merge(
    terminal_velocity_features,
    on="TRANSACTION_ID",
    how="left",
    validate="one_to_one"
)

print("TERMINAL FEATURES ATTACHED")
print("=" * 50)

print("Rows:", len(features_df))

print("\nTerminal velocity NaN counts:")
print(
    features_df[
        [
            "TERMINAL_TX_COUNT_5M",
            "TERMINAL_TX_COUNT_1H",
            "TERMINAL_TX_COUNT_24H"
        ]
    ].isna().sum()
)

print("\nCurrent feature count:", len(features_df.columns))
```

```python
# ============================================================
# HISTORICAL AMOUNT BASELINES
# ============================================================

# -------------------------
# Customer amount history
# -------------------------

features_df["CUSTOMER_PREV_AMOUNT_SUM"] = (
    features_df
    .groupby("CUSTOMER_ID")["TX_AMOUNT"]
    .cumsum()
    - features_df["TX_AMOUNT"]
)

features_df["CUSTOMER_PREV_AMOUNT_COUNT"] = (
    features_df["CUSTOMER_TX_COUNT_BEFORE"]
)

features_df["CUSTOMER_AVG_AMOUNT_BEFORE"] = np.where(
    features_df["CUSTOMER_PREV_AMOUNT_COUNT"] > 0,
    features_df["CUSTOMER_PREV_AMOUNT_SUM"] /
    features_df["CUSTOMER_PREV_AMOUNT_COUNT"],
    np.nan
)

# -------------------------
# Terminal amount history
# -------------------------

features_df["TERMINAL_PREV_AMOUNT_SUM"] = (
    features_df
    .groupby("TERMINAL_ID")["TX_AMOUNT"]
    .cumsum()
    - features_df["TX_AMOUNT"]
)

features_df["TERMINAL_PREV_AMOUNT_COUNT"] = (
    features_df["TERMINAL_TX_COUNT_BEFORE"]
)

features_df["TERMINAL_AVG_AMOUNT_BEFORE"] = np.where(
    features_df["TERMINAL_PREV_AMOUNT_COUNT"] > 0,
    features_df["TERMINAL_PREV_AMOUNT_SUM"] /
    features_df["TERMINAL_PREV_AMOUNT_COUNT"],
    np.nan
)

# -------------------------
# Amount deviation
# -------------------------

features_df["CUSTOMER_AMOUNT_DEVIATION"] = (
    features_df["TX_AMOUNT"] -
    features_df["CUSTOMER_AVG_AMOUNT_BEFORE"]
)

features_df["TERMINAL_AMOUNT_DEVIATION"] = (
    features_df["TX_AMOUNT"] -
    features_df["TERMINAL_AVG_AMOUNT_BEFORE"]
)

print("HISTORICAL AMOUNT FEATURES")
print("=" * 50)

print(
    features_df[
        [
            "CUSTOMER_ID",
            "TX_AMOUNT",
            "CUSTOMER_AVG_AMOUNT_BEFORE",
            "CUSTOMER_AMOUNT_DEVIATION",
            "TERMINAL_ID",
            "TERMINAL_AVG_AMOUNT_BEFORE",
            "TERMINAL_AMOUNT_DEVIATION"
        ]
    ].head(20)
)

print("\nFeature count:", len(features_df.columns))
```

```python
print("AMOUNT BASELINE VALIDATION")
print("=" * 50)

amount_features = [
    "CUSTOMER_AVG_AMOUNT_BEFORE",
    "CUSTOMER_AMOUNT_DEVIATION",
    "TERMINAL_AVG_AMOUNT_BEFORE",
    "TERMINAL_AMOUNT_DEVIATION"
]

print("\nNaN counts:")
print(features_df[amount_features].isna().sum())

print("\nAmount baseline statistics:")
print(
    features_df[
        [
            "TX_AMOUNT",
            "CUSTOMER_AVG_AMOUNT_BEFORE",
            "TERMINAL_AVG_AMOUNT_BEFORE"
        ]
    ].describe()
)

print("\nSample transactions with history:")

sample_with_history = features_df[
    (features_df["CUSTOMER_TX_COUNT_BEFORE"] > 0) &
    (features_df["TERMINAL_TX_COUNT_BEFORE"] > 0)
][
    [
        "TRANSACTION_ID",
        "TX_AMOUNT",
        "CUSTOMER_AVG_AMOUNT_BEFORE",
        "CUSTOMER_AMOUNT_DEVIATION",
        "TERMINAL_AVG_AMOUNT_BEFORE",
        "TERMINAL_AMOUNT_DEVIATION"
    ]
].head(10)

print(sample_with_history)
```

```python
# ============================================================
# RELATIVE AMOUNT + SYSTEM ACTIVITY FEATURES
# ============================================================

# ------------------------------------------------------------
# 1. Relative customer amount
# ------------------------------------------------------------

features_df["CUSTOMER_AMOUNT_RATIO"] = (
    features_df["TX_AMOUNT"] /
    features_df["CUSTOMER_AVG_AMOUNT_BEFORE"]
)

# ------------------------------------------------------------
# 2. Relative terminal amount
# ------------------------------------------------------------

features_df["TERMINAL_AMOUNT_RATIO"] = (
    features_df["TX_AMOUNT"] /
    features_df["TERMINAL_AVG_AMOUNT_BEFORE"]
)

# ------------------------------------------------------------
# Handle invalid/infinite ratios
# ------------------------------------------------------------

features_df[
    [
        "CUSTOMER_AMOUNT_RATIO",
        "TERMINAL_AMOUNT_RATIO"
    ]
] = (
    features_df[
        [
            "CUSTOMER_AMOUNT_RATIO",
            "TERMINAL_AMOUNT_RATIO"
        ]
    ]
    .replace([np.inf, -np.inf], np.nan)
)

# ------------------------------------------------------------
# 3. System-wide transaction velocity
# ------------------------------------------------------------

system_velocity_base = features_df[
    ["TRANSACTION_ID", "TX_DATETIME"]
].copy()

system_velocity_base = system_velocity_base.sort_values(
    ["TX_DATETIME", "TRANSACTION_ID"]
)

timestamps_ns = (
    system_velocity_base["TX_DATETIME"]
    .values
    .astype("datetime64[ns]")
    .astype(np.int64)
)

for window, column in [
    (5 * 60, "SYSTEM_TX_COUNT_5M"),
    (60 * 60, "SYSTEM_TX_COUNT_1H"),
    (24 * 60 * 60, "SYSTEM_TX_COUNT_24H")
]:

    window_ns = window * 1_000_000_000

    left_boundary = timestamps_ns - window_ns

    left_indices = np.searchsorted(
        timestamps_ns,
        left_boundary,
        side="left"
    )

    right_indices = np.searchsorted(
        timestamps_ns,
        timestamps_ns,
        side="left"
    )

    system_velocity_base[column] = (
        right_indices - left_indices
    ).astype(int)

# ------------------------------------------------------------
# Merge system features
# ------------------------------------------------------------

system_velocity_features = system_velocity_base[
    [
        "TRANSACTION_ID",
        "SYSTEM_TX_COUNT_5M",
        "SYSTEM_TX_COUNT_1H",
        "SYSTEM_TX_COUNT_24H"
    ]
]

features_df = features_df.merge(
    system_velocity_features,
    on="TRANSACTION_ID",
    how="left",
    validate="one_to_one"
)

# ------------------------------------------------------------
# Validation
# ------------------------------------------------------------

print("RELATIVE AMOUNT + SYSTEM ACTIVITY")
print("=" * 50)

print("Rows:", len(features_df))

print("\nRelative amount NaNs:")
print(
    features_df[
        [
            "CUSTOMER_AMOUNT_RATIO",
            "TERMINAL_AMOUNT_RATIO"
        ]
    ].isna().sum()
)

print("\nSystem velocity NaNs:")
print(
    features_df[
        [
            "SYSTEM_TX_COUNT_5M",
            "SYSTEM_TX_COUNT_1H",
            "SYSTEM_TX_COUNT_24H"
        ]
    ].isna().sum()
)

print("\nSystem velocity maximum:")
print(
    features_df[
        [
            "SYSTEM_TX_COUNT_5M",
            "SYSTEM_TX_COUNT_1H",
            "SYSTEM_TX_COUNT_24H"
        ]
    ].max()
)

print("\nCurrent feature count:", len(features_df.columns))
```

```python
# ============================================================
# PHASE 3.1 — PREPARE ML DATA
# ============================================================

# Temporal split
train_model = model_df[
    model_df["TX_DATETIME"] < "2018-08-01"
].copy()

val_model = model_df[
    (model_df["TX_DATETIME"] >= "2018-08-01") &
    (model_df["TX_DATETIME"] < "2018-09-01")
].copy()

test_model = model_df[
    model_df["TX_DATETIME"] >= "2018-09-01"
].copy()

# Feature matrices
X_train = train_model[model_features]
y_train = train_model[target]

X_val = val_model[model_features]
y_val = val_model[target]

X_test = test_model[model_features]
y_test = test_model[target]

print("PHASE 3.1 — ML DATA PREPARATION")
print("=" * 55)

print(f"X_train : {X_train.shape}")
print(f"y_train : {y_train.shape}")

print(f"X_val   : {X_val.shape}")
print(f"y_val   : {y_val.shape}")

print(f"X_test  : {X_test.shape}")
print(f"y_test  : {y_test.shape}")

print("\nFraud distribution")
print("-" * 55)

print(
    f"Train      : {y_train.sum():,} fraud / "
    f"{len(y_train):,} "
    f"({y_train.mean():.4%})"
)

print(
    f"Validation : {y_val.sum():,} fraud / "
    f"{len(y_val):,} "
    f"({y_val.mean():.4%})"
)

print(
    f"Test       : {y_test.sum():,} fraud / "
    f"{len(y_test):,} "
    f"({y_test.mean():.4%})"
)

print("\nMissing values")
print("-" * 55)

print("Train:", X_train.isna().sum().sum())
print("Val  :", X_val.isna().sum().sum())
print("Test :", X_test.isna().sum().sum())
```

```python
# ============================================================
# PHASE 3.2 — LOGISTIC REGRESSION BASELINE
# ============================================================

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

print("Training Logistic Regression baseline...")
print("=" * 55)

baseline_model = Pipeline([
    (
        "scaler",
        StandardScaler()
    ),
    (
        "classifier",
        LogisticRegression(
            class_weight="balanced",
            max_iter=1000,
            solver="lbfgs",
            random_state=42
        )
    )
])

baseline_model.fit(X_train, y_train)

print("Training complete.")
```

```python
# ============================================================
# PHASE 3.3 — BASELINE VALIDATION
# ============================================================

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    average_precision_score,
    roc_auc_score,
    confusion_matrix
)

# Fraud probabilities
val_prob = baseline_model.predict_proba(X_val)[:, 1]

# Default threshold
threshold = 0.5
val_pred = (val_prob >= threshold).astype(int)

# Metrics
precision = precision_score(y_val, val_pred, zero_division=0)
recall = recall_score(y_val, val_pred, zero_division=0)
f1 = f1_score(y_val, val_pred, zero_division=0)
pr_auc = average_precision_score(y_val, val_prob)
roc_auc = roc_auc_score(y_val, val_prob)

tn, fp, fn, tp = confusion_matrix(
    y_val,
    val_pred
).ravel()

print("LOGISTIC REGRESSION — VALIDATION")
print("=" * 55)

print(f"Threshold : {threshold}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")
print(f"PR-AUC    : {pr_auc:.4f}")
print(f"ROC-AUC   : {roc_auc:.4f}")

print("\nConfusion Matrix")
print("-" * 55)
print(f"True Negatives  : {tn:,}")
print(f"False Positives  : {fp:,}")
print(f"False Negatives  : {fn:,}")
print(f"True Positives   : {tp:,}")

print("\nPredicted fraud:")
print(f"{val_pred.sum():,}")
```

```python
# ============================================================
# PHASE 3.4 — THRESHOLD ANALYSIS
# ============================================================

thresholds = np.arange(0.10, 0.96, 0.05)

threshold_results = []

for threshold in thresholds:

    predictions = (val_prob >= threshold).astype(int)

    precision = precision_score(
        y_val,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_val,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_val,
        predictions,
        zero_division=0
    )

    tn, fp, fn, tp = confusion_matrix(
        y_val,
        predictions
    ).ravel()

    threshold_results.append({
        "threshold": threshold,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "false_positives": fp,
        "false_negatives": fn,
        "true_positives": tp
    })

threshold_results = pd.DataFrame(threshold_results)

print("THRESHOLD ANALYSIS")
print("=" * 80)

print(
    threshold_results.to_string(
        index=False,
        formatters={
            "threshold": "{:.2f}".format,
            "precision": "{:.4f}".format,
            "recall": "{:.4f}".format,
            "f1": "{:.4f}".format
        }
    )
)
```

```python
# ============================================================
# THRESHOLD RESULT SANITY CHECK
# ============================================================

check_thresholds = [0.10, 0.20, 0.50, 0.75, 0.80, 0.85, 0.90]

for threshold in check_thresholds:

    predictions = (val_prob >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_val,
        predictions
    ).ravel()

    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0

    print(
        f"Threshold {threshold:.2f} | "
        f"Precision: {precision:.6f} | "
        f"Recall: {recall:.6f} | "
        f"FP: {fp:,} | "
        f"TP: {tp:,}"
    )
```
